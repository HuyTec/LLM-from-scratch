# Data Preparation: Chuẩn bị dữ liệu
Kiểm tra, làm sạch, chuẩn hóa, tổ chức và chia dữ liệu.

## 0. Load dữ liệu

```python
from pathlib import Path

text = Path("../data/raw/input.txt").read_text(encoding="utf-8")
print(f'Length: {len(text)}')
print(text[:1000]) # 1000 ký tự đầu tiên
```

## 1. Làm sạch và chuẩn hóa văn bản
- Giữ nguyên văn phong, dấu câu, hoa thường,...
- Chỉ chuẩn hóa những yếu tố gây ra sự thiếu nhất quán trong cách biểu diễn văn bản: 

```python
import re
import unicodedata

def clean_text(text):

    # Loại bỏ các ký tự không phải chữ cái, số, dấu câu và khoảng trắng
    text = re.sub(r'[^a-zA-Z0-9\s.,!?;:\'\"-]', '', text)

    # Chuẩn hóa Unicode
    text = unicodedata.normalize('NFKC', text)

    # Giới hạn 2 lần khoảng trắng liên tiếp thành 1 khoảng trắng
    text = re.sub(r'\s{2,}', ' ', text)

    #Giới hạn chỉ tồn tại nhiều nhất hai lần xuống dòng liên tiếp
    text = re.sub(r'\n{3,}', '\n\n', text)

    return text

print(f'Length after cleaning: {len(clean_text(text))}')
```

```python
# Example usage
sample_text = "This is a sample text with some unwanted characters: @#$%^&*() and multiple     spaces.\n\n\nAnd some new lines."
cleaned_text = clean_text(sample_text)
print(f'Cleaned sample text: "{cleaned_text}"')
```
